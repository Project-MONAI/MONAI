# Clinical DICOM CT and MRI Preprocessing with MONAI
This notebook demonstrates inference-time preprocessing pipelines for CT and MRI DICOM series using MONAI, suitable for PACS/RIS workflows in hospital radiology environments.
**Note:** This notebook focuses on preprocessing only and excludes training or patient-identifiable data.

## Import Libraries

```python
from monai.transforms import (
    LoadImage,
    EnsureChannelFirst,
    ScaleIntensityRange,
    NormalizeIntensity,
    Compose
)
import numpy as np
```

## Define Preprocessing Pipelines

```python
def get_ct_preprocessing_pipeline():
    return Compose([
        LoadImage(image_only=True),
        EnsureChannelFirst(),
        ScaleIntensityRange(a_min=-1000, a_max=400, b_min=0.0, b_max=1.0, clip=True)
    ])

def get_mri_preprocessing_pipeline():
    return Compose([
        LoadImage(image_only=True),
        EnsureChannelFirst(),
        NormalizeIntensity(nonzero=True)
    ])
```

## Preprocessing Function

```python
def preprocess_dicom_series(dicom_path, modality):
    modality = modality.upper()
    if modality == 'CT':
        transform = get_ct_preprocessing_pipeline()
    elif modality == 'MRI':
        transform = get_mri_preprocessing_pipeline()
    else:
        raise ValueError("Unsupported modality. Use 'CT' or 'MRI'.")
    image = transform(dicom_path)
    return image
```

## Example Usage

```python
# Replace these paths with your own DICOM series paths
ct_dicom_path = '/path/to/ct/dicom/series'
mri_dicom_path = '/path/to/mri/dicom/series'

ct_image = preprocess_dicom_series(ct_dicom_path, 'CT')
mri_image = preprocess_dicom_series(mri_dicom_path, 'MRI')

print('CT image shape:', ct_image.shape)
print('MRI image shape:', mri_image.shape)
```
